## A --- TFD d'un cosinus

### A5 : cas exact, nu0 = 3/8

```python
N = 8
nu0 = 3 / 8
n = np.arange(N)
x = np.cos(2 * np.pi * nu0 * n)
X = tfd(x)

plot_signal_discret(x, title=f"x[n], nu0={nu0:.3f}")
plt.show()
plot_tfd(X, title=f"|X[k]|, nu0={nu0:.3f}")
plt.show()
```

### A6 : periodicite de X[k], k0=3, N=16

```python
N = 16
k0 = 3
nu0 = k0 / N
n = np.arange(N)
x = np.cos(2 * np.pi * nu0 * n)

k_range = np.arange(0, 2 * N)
X_elargi = tfd_k(x, k_range)
plot_tfd_k(k_range, X_elargi, title="|X[k]| pour k=0,...,2N-1")
plt.show()
```

### A7 : cas non exact, nu0 = 2.5/8

```python
N = 8
n = np.arange(N)
nu0 = 2.5 / 8
x = np.cos(2 * np.pi * nu0 * n)
X = tfd(x)

plot_signal_discret(x, title=f"x[n], nu0={nu0:.3f}")
plt.show()
plot_tfd(X, title=f"|X[k]|, nu0={nu0:.3f}")
plt.show()
```

### A8 : balayage continu entre deux rangs

```python
for nu0 in [3 / 8, 3.2 / 8, 3.5 / 8, 3.8 / 8, 4 / 8]:  # <- modifiez cette liste et ré-exécutez
    x = np.cos(2 * np.pi * nu0 * n)
    X = tfd(x)
    plot_tfd(X, title=f"nu0 = {nu0:.3f}")
    plt.show()
```

## C --- Compléter par des zéros (zero-padding)

### C1 : cosinus non exact (A4), complété jusqu'à N=128

```python
N0 = 8
nu0 = 1 / 8
n0 = np.arange(N0)
x = np.cos(2 * np.pi * nu0 * n0)

xp = zero_pad(x, 32)
Xp = tfd(xp)
plot_tfd(Xp, title="|X[k]|, cosinus complété à N=128")
plt.show()
```

### C2 : carré non exact (B3), complété jusqu'à N=512
