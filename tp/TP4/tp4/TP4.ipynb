## B --- Convolution avec une porte : moyenne glissante

### B1 : sur papier

### B2. B3. B4. B5. 

```python
def x_bruite(t,f=1):


    return (np.sin(2 * np.pi * f*t)
                + 0.4 * np.sin(2 * np.pi * 17 * t + 0.3)
                + 0.3 * np.sin(2 * np.pi * 34 * t + 1.1))

plot_function(x_bruite, 0, 2, title="x(t)")
plt.show()

theta = 1/17   # <- modifiez et ré-exécutez (B2 : 0.02, B3 : 0.15, B4 : 0.5)
plot_function(lambda t: moving_average(x_bruite, t, theta), 0, 2,
              title=f"moyenne glissante, theta={theta}")
plt.show()
```

## C --- Autocorrélation

### C1 : sur papier

### C2 : un seul triangle

```python
tri = make_triangle(theta=1.0)

plot_function(tri.f, -2,2, title="Triangle")
plt.show()

plot_function(lambda tau: autocorrelation(tri, tau), -2, 2,
              title="R_x(tau) pour un triangle", xlabel="$\\tau$")
plt.show()
```

### C3 : deux triangles séparés de D=4

```python
from tp4_helpers import Signal

D = 4
tri1 = make_triangle(theta=1.0, t0=-D / 2)
tri2 = make_triangle(theta=1.0, t0=D / 2)
x2 = Signal(f=lambda t: tri1.f(t) + tri2.f(t),
            support=(-D / 2 - 1, D / 2 + 1), name="deux triangles")


plot_function(x2.f, -6,6, title="2 triangles")
plt.show()

plot_function(lambda tau: autocorrelation(x2, tau, tmin=-6, tmax=6), -6, 6,
              title="R_x(tau), deux triangles", xlabel="tau")
plt.show()
```
